11-masala

Massivning shape ni o‘zgartirmasdan uning nusxasini (copy) yarat.

```python
import numpy as np

original_arr = np.array([[1, 2], [3, 4]])
print("Original array:\n", original_arr)
print("Shape of original array:", original_arr.shape)

# Create a copy without changing shape
copied_arr = original_arr.copy()
print("\nCopied array:\n", copied_arr)
print("Shape of copied array:", copied_arr.shape)

# Verify that they are different objects but have the same shape
print("\nAre they the same object?", original_arr is copied_arr)
print("Do they have the same shape?", original_arr.shape == copied_arr.shape)
```
