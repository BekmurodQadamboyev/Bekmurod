# **choice() ga oid masalalar (11–15)**

[1, 2, 3, 4, 5] ro‘yxatdan tasodifiy bitta element tanlang.

```python
import random
lst = [1, 2, 3, 4, 5]
tanlangan = random.choice(lst)
print(tanlangan)
```
